## 1. My lane as an ML task

This is a binary classification task. We are evaluating textual data (news articles) and sorting them into one of two discrete, mutually exclusive categories: authentic news or fake news. It is not a ranking or scoring problem because the primary operational goal is to assign a definitive label to support moderation or warning systems.

```python
# Define the task configuration and expected classes
task_type = "binary_classification"
target_classes = {
    0: "Authentic News",
    1: "Fake News"
}

print(f"Task: {task_type}")
print(f"Classes: {target_classes}")
```

## 2. Target or proxy

The target is a binary label (label column) where 1 indicates fake news and 0 indicates authentic news. This is an observed outcome derived from human fact-checkers or known historical source-level annotations, rather than a proxy rule (like predicting if an article will get a high click-through rate to guess if it's clickbait/fake).

```python
import pandas as pd
import numpy as np

# Simulating the target column to check class balance (replace with your actual df load)
# df = pd.read_csv('work/data/your_dataset.csv')
np.random.seed(42)
mock_targets = pd.Series(np.random.choice([0, 1], size=1000, p=[0.6, 0.4]))

# Check the distribution of our observed labels
distribution = mock_targets.value_counts(normalize=True).round(3)
print("Target label distribution:")
print(distribution)
```

## 3. Success metric
The primary success metric is the F1-score, supported by closely monitoring Precision. In this domain, false positives (flagging real news as fake) severely damage user trust, making high precision mandatory. However, we still need to catch a meaningful volume of misinformation (recall). The F1-score balances both. A macro F1-score above 0.85 indicates a strong, reliable baseline for decision-support.


```python
from sklearn.metrics import f1_score, precision_score

# Simulating a baseline prediction to test our metric functions
y_true = np.array([1, 0, 1, 1, 0, 0, 1, 0])
y_pred = np.array([1, 0, 0, 1, 0, 1, 1, 0])

current_f1 = f1_score(y_true, y_pred)
current_precision = precision_score(y_true, y_pred)

print(f"Target Metric (F1): > 0.85")
print(f"Dummy Baseline F1: {current_f1:.2f}")
print(f"Dummy Baseline Precision: {current_precision:.2f}")
```

## 5. Why ML beats a fixed rule here

Natural language is too high-dimensional and messy for static if/else rules. We cannot reliably classify fake news by simply matching keywords (e.g., if "shocking" in text: flag_as_fake()), because legitimate articles also use dramatic language, and misinformation constantly evolves its vocabulary to evade simple filters. A machine learning model detects subtle semantic patterns, word co-occurrences, and syntactic structures that human-coded rules would miss.

```python
# Demonstrating why a fixed rule fails
def bad_rule_based_classifier(text):
    # A naive rule that flags clickbait words
    suspicious_words = ['shocking', 'miracle', 'secret', 'you won\'t believe']
    text_lower = text.lower()

    for word in suspicious_words:
        if word in text_lower:
            return 1 # Fake
    return 0 # Authentic

# This authentic movie review gets falsely flagged by the rule
real_review = "The ending was shocking, a secret masterpiece of cinema."
prediction = bad_rule_based_classifier(real_review)

print(f"Text: '{real_review}'")
print(f"Rule-based prediction: {'Fake' if prediction == 1 else 'Authentic'} (Incorrect)")
print("Conclusion: Fixed rules suffer from high false-positive rates in natural language.")
```
